```python
# Configuration
VOCAB_SIZE = 8000
TOKENIZER_SAMPLE_SIZE = 200_000
SEED = 42

# Select deterministic training subset
tokenizer_data = train_dataset.select(
    range(TOKENIZER_SAMPLE_SIZE)
)

# Initialise BPE tokenizer
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# Byte-level pre-tokenisation
tokenizer.pre_tokenizer = ByteLevel(add_prefix_space=False)
tokenizer.decoder = ByteLevelDecoder()

# Define special tokens
trainer = BpeTrainer(
    vocab_size=VOCAB_SIZE,
    min_frequency=2,
    special_tokens=[
        "[UNK]",
        "[BOS]",
        "[EOS]",
        "[PAD]"
    ],
    initial_alphabet=ByteLevel.alphabet()
)

# Train using training stories only
def batch_iterator(batch_size=1000):
    for i in range(0, len(tokenizer_data), batch_size):
        yield tokenizer_data[i:i+batch_size]["text"]

tokenizer.train_from_iterator(
    batch_iterator(),
    trainer=trainer,
    length=len(tokenizer_data)
)

# Save tokenizer
os.makedirs("artifacts", exist_ok=True)
tokenizer.save("artifacts/tokenizer.json")

print("Tokenizer training completed!")
print("Vocabulary size:", tokenizer.get_vocab_size())
```

```python
BLOCK_SIZE = 256
BATCH_SIZE = 32

def get_batch(data, batch_size=BATCH_SIZE):
    """
    Generate random batches for autoregressive
    next-token prediction.
    """

    ix = torch.randint(
        len(data) - BLOCK_SIZE - 1,
        (batch_size,)
    )

    offsets = torch.arange(BLOCK_SIZE)

    x = data[ix[:, None] + offsets]
    y = data[ix[:, None] + offsets + 1]

    return (
        x.to(device),
        y.to(device)
    )


# Test the batch generator
xb, yb = get_batch(train_data)

print("Input shape:", xb.shape)
print("Target shape:", yb.shape)
print("Input dtype:", xb.dtype)
print("Target dtype:", yb.dtype)
print("GPU device:", xb.device)

assert xb.shape == (BATCH_SIZE, BLOCK_SIZE)
assert yb.shape == (BATCH_SIZE, BLOCK_SIZE)
assert torch.equal(xb[:, 1:], yb[:, :-1])
```

```python
%%writefile model.py

import torch
import torch.nn as nn
import torch.nn.functional as F


class CausalSelfAttention(nn.Module):
    """
    Multi-head causal self-attention.
    Each token can only attend to itself and previous tokens.
    """

    def __init__(self, n_embd, n_head, dropout=0.1):
        super().__init__()

        assert n_embd % n_head == 0

        self.n_head = n_head
        self.head_dim = n_embd // n_head

        self.qkv = nn.Linear(n_embd, 3 * n_embd)
        self.proj = nn.Linear(n_embd, n_embd)

        self.attn_dropout = dropout
        self.resid_dropout = nn.Dropout(dropout)

    def forward(self, x):
        B, T, C = x.shape

        # Project input into queries, keys and values
        qkv = self.qkv(x)

        q, k, v = qkv.split(C, dim=2)

        # Separate attention heads
        q = q.view(B, T, self.n_head, self.head_dim)
        k = k.view(B, T, self.n_head, self.head_dim)
        v = v.view(B, T, self.n_head, self.head_dim)

        q = q.transpose(1, 2)
        k = k.transpose(1, 2)
        v = v.transpose(1, 2)

        # Causal scaled dot-product attention
        y = F.scaled_dot_product_attention(
            q, k, v,
            is_causal=True,
            dropout_p=self.attn_dropout if self.training else 0.0
        )

        # Combine attention heads
        y = y.transpose(1, 2).contiguous()
        y = y.view(B, T, C)

        return self.resid_dropout(self.proj(y))


class FeedForward(nn.Module):
    """
    Position-wise feed-forward network.
    """

    def __init__(self, n_embd, dropout=0.1):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd),
            nn.GELU(),
            nn.Linear(4 * n_embd, n_embd),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        return self.net(x)


class TransformerBlock(nn.Module):
    """
    Pre-LayerNorm Transformer block.
    """

    def __init__(self, n_embd, n_head, dropout=0.1):
        super().__init__()

        self.ln1 = nn.LayerNorm(n_embd)
        self.attn = CausalSelfAttention(
            n_embd, n_head, dropout
        )

        self.ln2 = nn.LayerNorm(n_embd)
        self.ffn = FeedForward(n_embd, dropout)

    def forward(self, x):
        # Residual connections
        x = x + self.attn(self.ln1(x))
        x = x + self.ffn(self.ln2(x))

        return x


class GPTLanguageModel(nn.Module):
    """
    Decoder-only GPT language model.
    """

    def __init__(
        self,
        vocab_size=8000,
        n_embd=256,
        n_layer=4,
        n_head=4,
        block_size=256,
        dropout=0.1
    ):
        super().__init__()

        self.block_size = block_size

        # Token and positional embeddings
        self.token_embedding = nn.Embedding(
            vocab_size, n_embd
        )

        self.position_embedding = nn.Embedding(
            block_size, n_embd
        )

        self.dropout = nn.Dropout(dropout)

        # Transformer blocks
        self.blocks = nn.Sequential(
            *[
                TransformerBlock(n_embd, n_head, dropout)
                for _ in range(n_layer)
            ]
        )

        self.ln_f = nn.LayerNorm(n_embd)

        # Language modelling output head
        self.lm_head = nn.Linear(
            n_embd, vocab_size, bias=False
        )

        # Tie token embeddings and output projection weights
        self.lm_head.weight = self.token_embedding.weight

        # Initialise parameters
        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

            if module.bias is not None:
                nn.init.zeros_(module.bias)

        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx, targets=None):
        B, T = idx.shape

        assert T <= self.block_size

        # Token embeddings
        tok_emb = self.token_embedding(idx)

        # Positional embeddings
        positions = torch.arange(
            T, device=idx.device
        )

        pos_emb = self.position_embedding(positions)

        x = self.dropout(tok_emb + pos_emb)

        # Transformer
        x = self.blocks(x)
        x = self.ln_f(x)

        # Vocabulary logits
        logits = self.lm_head(x)

        loss = None

        if targets is not None:
            loss = F.cross_entropy(
                logits.reshape(-1, logits.size(-1)),
                targets.reshape(-1)
            )

        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens, temperature=1.0):
        self.eval()

        for _ in range(max_new_tokens):

            # Keep only the most recent context
            idx_cond = idx[:, -self.block_size:]

            logits, _ = self(idx_cond)

            # Select final-position logits
            logits = logits[:, -1, :]

            # Temperature scaling
            logits = logits / temperature

            # Sample next token
            probs = F.softmax(logits, dim=-1)

            next_token = torch.multinomial(
                probs, num_samples=1
            )

            idx = torch.cat((idx, next_token), dim=1)

        return idx
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import time
import math
import os

# Training configuration
BATCH_SIZE = 32
BLOCK_SIZE = 256

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 0.1

MAX_ITERS = 500
EVAL_INTERVAL = 100
EVAL_ITERS = 30

WARMUP_ITERS = 50

GRAD_CLIP = 1.0

# Optimiser
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    betas=(0.9, 0.95),
    weight_decay=WEIGHT_DECAY
)

# Mixed precision
scaler = torch.amp.GradScaler("cuda")

os.makedirs("checkpoints", exist_ok=True)

print("Training configuration ready.")
```

```python
def get_lr(iteration):
    if iteration < WARMUP_ITERS:
        return LEARNING_RATE * (iteration + 1) / WARMUP_ITERS

    progress = (
        iteration - WARMUP_ITERS
    ) / (MAX_ITERS - WARMUP_ITERS)

    progress = min(progress, 1.0)

    return LEARNING_RATE * 0.5 * (
        1.0 + math.cos(math.pi * progress)
    )
```

```python
@torch.no_grad()
def estimate_loss():
    model.eval()

    results = {}

    for name, data in [
        ("train", train_data),
        ("val", val_data)
    ]:
        losses = torch.zeros(EVAL_ITERS)

        for k in range(EVAL_ITERS):
            x, y = get_batch(data)

            with torch.amp.autocast(
                "cuda",
                dtype=torch.float16
            ):
                _, loss = model(x, y)

            losses[k] = loss.item()

        results[name] = losses.mean().item()

    model.train()

    return results
```

```python
model.train()

training_history = []

start_time = time.time()

for iteration in range(MAX_ITERS):

    # Update learning rate
    lr = get_lr(iteration)

    for param_group in optimizer.param_groups:
        param_group["lr"] = lr

    # Sample batch
    xb, yb = get_batch(train_data)

    # Clear gradients
    optimizer.zero_grad(set_to_none=True)

    # Forward pass
    with torch.amp.autocast(
        "cuda",
        dtype=torch.float16
    ):
        logits, loss = model(xb, yb)

    # Backward pass
    scaler.scale(loss).backward()

    # Unscale before gradient clipping
    scaler.unscale_(optimizer)

    torch.nn.utils.clip_grad_norm_(
        model.parameters(),
        GRAD_CLIP
    )

    # Optimiser update
    scaler.step(optimizer)
    scaler.update()

    # Record progress
    if iteration % EVAL_INTERVAL == 0 or iteration == MAX_ITERS - 1:

        results = estimate_loss()

        elapsed = time.time() - start_time

        training_history.append({
            "iteration": iteration,
            "train_loss": results["train"],
            "val_loss": results["val"],
            "learning_rate": lr,
            "elapsed_seconds": elapsed
        })

        print(
            f"Step {iteration:4d} | "
            f"Train loss: {results['train']:.4f} | "
            f"Val loss: {results['val']:.4f} | "
            f"LR: {lr:.6f} | "
            f"Time: {elapsed:.1f}s"
        )

# Save checkpoint
torch.save({
    "model_state_dict": model.state_dict(),
    "optimizer_state_dict": optimizer.state_dict(),
    "config": {
        "vocab_size": 8000,
        "n_embd": 256,
        "n_layer": 4,
        "n_head": 4,
        "block_size": 256
    },
    "training_history": training_history
}, "checkpoints/gpt_baseline_pilot.pt")

print("\nPilot training completed!")
print("Total time:", round(time.time() - start_time, 2), "seconds")
```

```python

import torch
import torch.nn.functional as F
import time
import math
import os

# ==========================================
# 1. EXPERIMENT CONFIGURATION
# ==========================================

MAX_ITERS = 5000
BATCH_SIZE = 32
BLOCK_SIZE = 256

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 0.1

WARMUP_ITERS = 200
EVAL_INTERVAL = 500
EVAL_ITERS = 50

GRAD_CLIP = 1.0

# ==========================================
# 2. INITIALISE A FRESH MODEL
# ==========================================

model = GPTLanguageModel(
    vocab_size=8000,
    n_embd=256,
    n_layer=4,
    n_head=4,
    block_size=256,
    dropout=0.1
).to(device)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    betas=(0.9, 0.95),
    weight_decay=WEIGHT_DECAY
)

scaler = torch.amp.GradScaler("cuda")

# ==========================================
# 3. LEARNING RATE SCHEDULE
# ==========================================

def get_lr(iteration):
    if iteration < WARMUP_ITERS:
        return LEARNING_RATE * (iteration + 1) / WARMUP_ITERS

    progress = (
        iteration - WARMUP_ITERS
    ) / (MAX_ITERS - WARMUP_ITERS)

    progress = min(progress, 1.0)

    return LEARNING_RATE * 0.5 * (
        1 + math.cos(math.pi * progress)
    )

# ==========================================
# 4. VALIDATION FUNCTION
# ==========================================

@torch.no_grad()
def evaluate_model():
    model.eval()

    results = {}

    for name, data in [
        ("train", train_data),
        ("val", val_data)
    ]:
        losses = []

        for _ in range(EVAL_ITERS):
            x, y = get_batch(data)

            with torch.amp.autocast(
                "cuda",
                dtype=torch.float16
            ):
                _, loss = model(x, y)

            losses.append(loss.item())

        results[name] = sum(losses) / len(losses)

    model.train()
    return results

# ==========================================
# 5. TRAINING
# ==========================================

os.makedirs("checkpoints", exist_ok=True)

training_history_5000 = []

model.train()

torch.cuda.reset_peak_memory_stats()
torch.cuda.synchronize()

start_time = time.time()

for iteration in range(MAX_ITERS):

    # Update learning rate
    lr = get_lr(iteration)

    for param_group in optimizer.param_groups:
        param_group["lr"] = lr

    # Get batch
    xb, yb = get_batch(train_data)

    # Clear gradients
    optimizer.zero_grad(set_to_none=True)

    # Forward pass
    with torch.amp.autocast(
        "cuda",
        dtype=torch.float16
    ):
        logits, loss = model(xb, yb)

    # Backward pass
    scaler.scale(loss).backward()

    scaler.unscale_(optimizer)

    torch.nn.utils.clip_grad_norm_(
        model.parameters(),
        GRAD_CLIP
    )

    scaler.step(optimizer)
    scaler.update()

    # Evaluate periodically
    if iteration % EVAL_INTERVAL == 0 or iteration == MAX_ITERS - 1:

        results = evaluate_model()

        torch.cuda.synchronize()
        elapsed = time.time() - start_time

        record = {
            "iteration": iteration,
            "train_loss": results["train"],
            "val_loss": results["val"],
            "learning_rate": lr,
            "elapsed_seconds": elapsed
        }

        training_history_5000.append(record)

        print(
            f"Step {iteration:5d} | "
            f"Train loss: {results['train']:.4f} | "
            f"Val loss: {results['val']:.4f} | "
            f"LR: {lr:.6f} | "
            f"Time: {elapsed:.1f}s",
            flush=True
        )

# ==========================================
# 6. SAVE CHECKPOINT
# ==========================================

torch.cuda.synchronize()

total_time = time.time() - start_time

peak_memory = (
    torch.cuda.max_memory_allocated() / 1e9
)

checkpoint = {
    "model_state_dict": model.state_dict(),
    "optimizer_state_dict": optimizer.state_dict(),
    "training_history": training_history_5000,
    "config": {
        "vocab_size": 8000,
        "n_embd": 256,
        "n_layer": 4,
        "n_head": 4,
        "block_size": 256,
        "max_iters": MAX_ITERS,
        "batch_size": BATCH_SIZE,
        "learning_rate": LEARNING_RATE
    }
}

torch.save(
    checkpoint,
    "checkpoints/gpt_baseline_5000.pt"
)

print("\n" + "=" * 45)
print("EXTENDED BASELINE COMPLETED")
print("=" * 45)

print("Total training time:", round(total_time, 2), "seconds")
print("Peak GPU memory:", round(peak_memory, 2), "GB")
print("Final training loss:",
      training_history_5000[-1]["train_loss"])
print("Final validation loss:",
      training_history_5000[-1]["val_loss"])
print("Checkpoint saved successfully.")
```

```python

# =====================================================
# EXPERIMENT E2: DATA SCALING
# Train D1, D2 and D3 under controlled conditions
# =====================================================

import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import time
import math
import os
from collections import deque

# -----------------------------------------------------
# 1. EXPERIMENT CONFIGURATION
# -----------------------------------------------------

MAX_ITERS = 5000
BATCH_SIZE = 32
BLOCK_SIZE = 256

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 0.1

WARMUP_ITERS = 200
EVAL_INTERVAL = 500
EVAL_ITERS = 50

GRAD_CLIP = 1.0
SEED = 1234

# Use the same configurations for all datasets
DATA_CONFIGS = {
    "D1_10k": train_streams[10_000],
    "D2_50k": train_streams[50_000],
    "D3_100k": train_streams[100_000]
}

os.makedirs("checkpoints/data_scaling", exist_ok=True)

# -----------------------------------------------------
# 2. FIXED VALIDATION BATCHES
# -----------------------------------------------------

# Every model will be evaluated on exactly the same
# validation batches.

val_generator = torch.Generator()
val_generator.manual_seed(999)

validation_batches = []

for _ in range(EVAL_ITERS):

    ix = torch.randint(
        len(val_data) - BLOCK_SIZE - 1,
        (BATCH_SIZE,),
        generator=val_generator
    )

    offsets = torch.arange(BLOCK_SIZE)

    x = val_data[ix[:, None] + offsets]
    y = val_data[ix[:, None] + offsets + 1]

    validation_batches.append((
        x.to(device),
        y.to(device)
    ))

print("Fixed validation batches prepared:", len(validation_batches))

# -----------------------------------------------------
# 3. HELPER FUNCTIONS
# -----------------------------------------------------

def sample_training_batch(data, generator):

    ix = torch.randint(
        len(data) - BLOCK_SIZE - 1,
        (BATCH_SIZE,),
        generator=generator
    )

    offsets = torch.arange(BLOCK_SIZE)

    x = data[ix[:, None] + offsets]
    y = data[ix[:, None] + offsets + 1]

    return x.to(device), y.to(device)


def get_learning_rate(iteration):

    if iteration < WARMUP_ITERS:
        return LEARNING_RATE * (
            iteration + 1
        ) / WARMUP_ITERS

    progress = (
        iteration - WARMUP_ITERS
    ) / (MAX_ITERS - WARMUP_ITERS)

    progress = min(progress, 1.0)

    return LEARNING_RATE * 0.5 * (
        1 + math.cos(math.pi * progress)
    )


@torch.no_grad()
def evaluate_validation(model):

    model.eval()

    losses = []

    for x, y in validation_batches:

        with torch.amp.autocast(
            "cuda",
            dtype=torch.float16
        ):
            _, loss = model(x, y)

        losses.append(loss.item())

    model.train()

    return float(np.mean(losses))


# -----------------------------------------------------
# 4. TRAIN ALL THREE MODELS
# -----------------------------------------------------

all_results = {}
summary_results = []

for experiment_name, training_data in DATA_CONFIGS.items():

    print("\n" + "=" * 60)
    print("STARTING:", experiment_name)
    print("Training tokens:", len(training_data))
    print("=" * 60)

    # Identical initialisation seed for each model
    torch.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)

    # Fresh model
    model = GPTLanguageModel(
        vocab_size=8000,
        n_embd=256,
        n_layer=4,
        n_head=4,
        block_size=256,
        dropout=0.1
    ).to(device)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        betas=(0.9, 0.95),
        weight_decay=WEIGHT_DECAY
    )

    scaler = torch.amp.GradScaler("cuda")

    # Independent but identically seeded data sampler
    train_generator = torch.Generator()
    train_generator.manual_seed(SEED + 1)

    model.train()

    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

    start_time = time.time()

    history = []
    recent_losses = deque(maxlen=100)

    # -------------------------------------------------
    # TRAINING LOOP
    # -------------------------------------------------

    for iteration in range(MAX_ITERS):

        lr = get_learning_rate(iteration)

        for param_group in optimizer.param_groups:
            param_group["lr"] = lr

        xb, yb = sample_training_batch(
            training_data,
            train_generator
        )

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast(
            "cuda",
            dtype=torch.float16
        ):
            logits, loss = model(xb, yb)

        scaler.scale(loss).backward()

        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            GRAD_CLIP
        )

        scaler.step(optimizer)
        scaler.update()

        recent_losses.append(loss.item())

        # -------------------------------------------------
        # PERIODIC EVALUATION
        # -------------------------------------------------

        if (
            iteration % EVAL_INTERVAL == 0
            or iteration == MAX_ITERS - 1
        ):

            val_loss = evaluate_validation(model)

            torch.cuda.synchronize()

            elapsed = time.time() - start_time

            train_loss = float(np.mean(recent_losses))

            record = {
                "iteration": iteration,
                "train_loss": train_loss,
                "val_loss": val_loss,
                "learning_rate": lr,
                "elapsed_seconds": elapsed
            }

            history.append(record)

            print(
                f"{experiment_name} | "
                f"Step {iteration:5d} | "
                f"Train: {train_loss:.4f} | "
                f"Val: {val_loss:.4f} | "
                f"Time: {elapsed:.1f}s",
                flush=True
            )

    # -------------------------------------------------
    # SAVE RESULTS
    # -------------------------------------------------

    torch.cuda.synchronize()

    total_time = time.time() - start_time

    peak_memory = (
        torch.cuda.max_memory_allocated() / 1e9
    )

    final_val_loss = history[-1]["val_loss"]

    # Save individual model checkpoint
    checkpoint = {
        "model_state_dict": model.state_dict(),
        "config": {
            "dataset": experiment_name,
            "training_tokens": len(training_data),
            "max_iters": MAX_ITERS,
            "batch_size": BATCH_SIZE,
            "block_size": BLOCK_SIZE,
            "seed": SEED
        },
        "history": history
    }

    torch.save(
        checkpoint,
        f"checkpoints/data_scaling/{experiment_name}.pt"
    )

    all_results[experiment_name] = history

    summary_results.append({
        "Dataset": experiment_name,
        "Training tokens": len(training_data),
        "Final train loss": history[-1]["train_loss"],
        "Final val loss": final_val_loss,
        "Val perplexity": math.exp(final_val_loss),
        "Training time (s)": total_time,
        "Peak GPU memory (GB)": peak_memory
    })

    print("\nCompleted:", experiment_name)
    print("Final validation loss:", final_val_loss)
    print("Training time:", round(total_time, 2), "seconds")
    print("Peak memory:", round(peak_memory, 2), "GB")

    del model, optimizer, scaler
    torch.cuda.empty_cache()

# -----------------------------------------------------
# 5. FINAL COMPARISON TABLE
# -----------------------------------------------------

results_df = pd.DataFrame(summary_results)

print("\n" + "=" * 70)
print("EXPERIMENT E2: FINAL RESULTS")
print("=" * 70)

display(results_df)

results_df.to_csv(
    "artifacts/data_scaling_results.csv",
    index=False
)

print("Results saved successfully.")
```

```python

# =====================================================
# EXPERIMENT E2 REPLICATION
# Three datasets x three random seeds
# =====================================================

import torch
import numpy as np
import pandas as pd
import time
import math
import os
import gc
from tqdm.auto import tqdm

# -----------------------------------------------------
# 1. CONFIGURATION
# -----------------------------------------------------

MAX_ITERS = 5000
BATCH_SIZE = 32
BLOCK_SIZE = 256

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 0.1

WARMUP_ITERS = 200
EVAL_INTERVAL = 500
EVAL_ITERS = 50

GRAD_CLIP = 1.0

SEEDS = [2027, 2028]

DATA_CONFIGS = {
    "D1_10k": train_streams[10_000],
    "D2_50k": train_streams[50_000],
    "D3_100k": train_streams[100_000]
}

os.makedirs("checkpoints/data_scaling", exist_ok=True)
os.makedirs("artifacts/data_scaling", exist_ok=True)

# -----------------------------------------------------
# 2. FIXED VALIDATION BATCHES
# -----------------------------------------------------

# Recreate the same validation batches used in E2.

val_generator = torch.Generator()
val_generator.manual_seed(999)

validation_batches = []

for _ in range(EVAL_ITERS):

    ix = torch.randint(
        len(val_data) - BLOCK_SIZE - 1,
        (BATCH_SIZE,),
        generator=val_generator
    )

    offsets = torch.arange(BLOCK_SIZE)

    x = val_data[ix[:, None] + offsets]
    y = val_data[ix[:, None] + offsets + 1]

    validation_batches.append((
        x.to(device),
        y.to(device)
    ))

print("Fixed validation batches:", len(validation_batches))

# -----------------------------------------------------
# 3. HELPER FUNCTIONS
# -----------------------------------------------------

def get_lr_rep(iteration):

    if iteration < WARMUP_ITERS:
        return LEARNING_RATE * (
            iteration + 1
        ) / WARMUP_ITERS

    progress = (
        iteration - WARMUP_ITERS
    ) / (MAX_ITERS - WARMUP_ITERS)

    progress = min(progress, 1.0)

    return LEARNING_RATE * 0.5 * (
        1 + math.cos(math.pi * progress)
    )


def sample_batch_rep(data, generator):

    ix = torch.randint(
        len(data) - BLOCK_SIZE - 1,
        (BATCH_SIZE,),
        generator=generator
    )

    offsets = torch.arange(BLOCK_SIZE)

    x = data[ix[:, None] + offsets]
    y = data[ix[:, None] + offsets + 1]

    return x.to(device), y.to(device)


@torch.no_grad()
def evaluate_rep(model):

    model.eval()

    losses = []

    for x, y in validation_batches:

        with torch.amp.autocast(
            "cuda",
            dtype=torch.float16
        ):
            _, loss = model(x, y)

        losses.append(loss.item())

    model.train()

    return float(np.mean(losses))


# -----------------------------------------------------
# 4. RUN REPLICATION EXPERIMENTS
# -----------------------------------------------------

replication_results = {}

replication_summary = []

for experiment_name, training_data in DATA_CONFIGS.items():

    for seed in SEEDS:

        run_name = f"{experiment_name}_seed{seed}"

        print("\n" + "=" * 65)
        print("STARTING:", run_name)
        print("Training tokens:", len(training_data))
        print("=" * 65)

        # Reproducible initialisation
        torch.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        np.random.seed(seed)

        # Fresh model
        model = GPTLanguageModel(
            vocab_size=8000,
            n_embd=256,
            n_layer=4,
            n_head=4,
            block_size=256,
            dropout=0.1
        ).to(device)

        optimizer = torch.optim.AdamW(
            model.parameters(),
            lr=LEARNING_RATE,
            betas=(0.9, 0.95),
            weight_decay=WEIGHT_DECAY
        )

        scaler = torch.amp.GradScaler("cuda")

        # Independent reproducible batch sampler
        train_generator = torch.Generator()
        train_generator.manual_seed(seed + 1)

        history = []
        recent_losses = []

        model.train()

        torch.cuda.reset_peak_memory_stats()
        torch.cuda.synchronize()

        start_time = time.time()

        # -------------------------------------------------
        # TRAINING LOOP
        # -------------------------------------------------

        for iteration in range(MAX_ITERS):

            lr = get_lr_rep(iteration)

            for param_group in optimizer.param_groups:
                param_group["lr"] = lr

            xb, yb = sample_batch_rep(
                training_data,
                train_generator
            )

            optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast(
                "cuda",
                dtype=torch.float16
            ):
                logits, loss = model(xb, yb)

            scaler.scale(loss).backward()

            scaler.unscale_(optimizer)

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                GRAD_CLIP
            )

            scaler.step(optimizer)
            scaler.update()

            recent_losses.append(loss.item())

            if (
                iteration % EVAL_INTERVAL == 0
                or iteration == MAX_ITERS - 1
            ):

                val_loss = evaluate_rep(model)

                torch.cuda.synchronize()
                elapsed = time.time() - start_time

                train_loss = float(
                    np.mean(recent_losses[-100:])
                )

                history.append({
                    "iteration": iteration,
                    "train_loss": train_loss,
                    "val_loss": val_loss,
                    "learning_rate": lr,
                    "elapsed_seconds": elapsed
                })

                print(
                    f"{run_name} | "
                    f"Step {iteration:5d} | "
                    f"Train: {train_loss:.4f} | "
                    f"Val: {val_loss:.4f} | "
                    f"Time: {elapsed:.1f}s",
                    flush=True
                )

        # -------------------------------------------------
        # SAVE CHECKPOINT AND RESULTS
        # -------------------------------------------------

        torch.cuda.synchronize()

        total_time = time.time() - start_time

        peak_memory = (
            torch.cuda.max_memory_allocated() / 1e9
        )

        final_train = history[-1]["train_loss"]
        final_val = history[-1]["val_loss"]

        checkpoint = {
            "model_state_dict": model.state_dict(),
            "config": {
                "dataset": experiment_name,
                "seed": seed,
                "training_tokens": len(training_data),
                "max_iters": MAX_ITERS,
                "batch_size": BATCH_SIZE,
                "block_size": BLOCK_SIZE
            },
            "history": history
        }

        torch.save(
            checkpoint,
            f"checkpoints/data_scaling/{run_name}.pt"
        )

        replication_results[run_name] = history

        replication_summary.append({
            "Dataset": experiment_name,
            "Seed": seed,
            "Training tokens": len(training_data),
            "Final train loss": final_train,
            "Final val loss": final_val,
            "Val perplexity": math.exp(final_val),
            "Training time (s)": total_time,
            "Peak GPU memory (GB)": peak_memory
        })

        print("\nCompleted:", run_name)
        print("Validation loss:", round(final_val, 5))
        print("Training time:", round(total_time, 2), "seconds")

        # Release resources
        del model, optimizer, scaler
        gc.collect()
        torch.cuda.empty_cache()

        # Save results after EVERY run
        pd.DataFrame(replication_summary).to_csv(
            "artifacts/data_scaling/replication_results.csv",
            index=False
        )

# -----------------------------------------------------
# 5. COMBINE WITH ORIGINAL SEED 1234
# -----------------------------------------------------

original_rows = []

for name, history in all_results.items():

    final = history[-1]

    original_rows.append({
        "Dataset": name,
        "Seed": 1234,
        "Training tokens": DATA_CONFIGS[name].numel(),
        "Final train loss": final["train_loss"],
        "Final val loss": final["val_loss"],
        "Val perplexity": math.exp(final["val_loss"])
    })

original_df = pd.DataFrame(original_rows)

replication_df = pd.DataFrame(replication_summary)

combined_df = pd.concat(
    [
        original_df,
        replication_df
    ],
    ignore_index=True
)

print("\n" + "=" * 65)
print("ALL EXPERIMENTAL RUNS")
print("=" * 65)

display(combined_df)

# -----------------------------------------------------
# 6. MEAN AND STANDARD DEVIATION
# -----------------------------------------------------

summary_df = combined_df.groupby("Dataset").agg(
    mean_val_loss=("Final val loss", "mean"),
    std_val_loss=("Final val loss", "std"),
    mean_perplexity=("Val perplexity", "mean"),
    std_perplexity=("Val perplexity", "std"),
    mean_train_loss=("Final train loss", "mean"),
    mean_training_time=("Training time (s)", "mean")
).reset_index()

print("\n" + "=" * 65)
print("E2 REPLICATION: MEAN +/- STANDARD DEVIATION")
print("=" * 65)

display(summary_df)

combined_df.to_csv(
    "artifacts/data_scaling/all_runs.csv",
    index=False
)

summary_df.to_csv(
    "artifacts/data_scaling/summary_statistics.csv",
    index=False
)

print("\nAll replication experiments completed!")
print("Results and checkpoints saved.")
```

```python

# =====================================================
# EXPERIMENT E3: MODEL SCALING UNDER FIXED TIME
# =====================================================

import torch
import numpy as np
import pandas as pd
import time
import math
import os
import gc

# -----------------------------------------------------
# 1. CONFIGURATION
# -----------------------------------------------------

TIME_BUDGET = 150       # seconds per model
BATCH_SIZE = 32
BLOCK_SIZE = 256

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 0.1
WARMUP_STEPS = 100
EVAL_INTERVAL = 500
EVAL_ITERS = 50
GRAD_CLIP = 1.0

SEED = 1234

# All models use the same training dataset
training_data = train_streams[100_000]

# Model configurations
MODEL_CONFIGS = {
    "Small_5M": {
        "n_embd": 256,
        "n_layer": 4,
        "n_head": 4
    },
    "Medium_14M": {
        "n_embd": 384,
        "n_layer": 6,
        "n_head": 6
    },
    "Large_30M": {
        "n_embd": 512,
        "n_layer": 8,
        "n_head": 8
    }
}

os.makedirs("checkpoints/model_scaling", exist_ok=True)
os.makedirs("artifacts/model_scaling", exist_ok=True)

# -----------------------------------------------------
# 2. FIXED VALIDATION BATCHES
# -----------------------------------------------------

# Use the same validation batches as E2 replication.
# If validation_batches already exists, retain it.

assert len(validation_batches) == EVAL_ITERS

# -----------------------------------------------------
# 3. HELPER FUNCTIONS
# -----------------------------------------------------

def evaluate_model_scaling(model):

    model.eval()
    losses = []

    with torch.no_grad():

        for x, y in validation_batches:

            with torch.amp.autocast(
                "cuda",
                dtype=torch.float16
            ):
                _, loss = model(x, y)

            losses.append(loss.item())

    model.train()

    return float(np.mean(losses))


# -----------------------------------------------------
# 4. TRAIN THREE MODEL SIZES
# -----------------------------------------------------

scaling_histories = {}
scaling_summary = []

for model_name, config in MODEL_CONFIGS.items():

    print("\n" + "=" * 65)
    print("STARTING:", model_name)
    print("=" * 65)

    # Reproducible initialisation
    torch.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)
    np.random.seed(SEED)

    model = GPTLanguageModel(
        vocab_size=8000,
        n_embd=config["n_embd"],
        n_layer=config["n_layer"],
        n_head=config["n_head"],
        block_size=BLOCK_SIZE,
        dropout=0.1
    ).to(device)

    n_params = sum(
        p.numel() for p in model.parameters()
    )

    print("Parameters:", f"{n_params:,}")

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        betas=(0.9, 0.95),
        weight_decay=WEIGHT_DECAY
    )

    scaler = torch.amp.GradScaler("cuda")

    train_generator = torch.Generator()
    train_generator.manual_seed(SEED + 1)

    model.train()

    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

    start_time = time.perf_counter()

    history = []
    iteration = 0
    recent_losses = []

    # -------------------------------------------------
    # FIXED WALL-CLOCK TRAINING BUDGET
    # -------------------------------------------------

    while time.perf_counter() - start_time < TIME_BUDGET:

        # Linear warmup, then constant learning rate
        if iteration < WARMUP_STEPS:
            lr = LEARNING_RATE * (
                iteration + 1
            ) / WARMUP_STEPS
        else:
            lr = LEARNING_RATE

        for group in optimizer.param_groups:
            group["lr"] = lr

        xb, yb = sample_batch_rep(
            training_data,
            train_generator
        )

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast(
            "cuda",
            dtype=torch.float16
        ):
            logits, loss = model(xb, yb)

        scaler.scale(loss).backward()

        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            GRAD_CLIP
        )

        scaler.step(optimizer)
        scaler.update()

        recent_losses.append(loss.item())
        iteration += 1

        # Periodic validation
        if iteration % EVAL_INTERVAL == 0:

            val_loss = evaluate_model_scaling(model)

            torch.cuda.synchronize()

            elapsed = time.perf_counter() - start_time

            record = {
                "iteration": iteration,
                "train_loss": float(
                    np.mean(recent_losses[-100:])
                ),
                "val_loss": val_loss,
                "elapsed_seconds": elapsed
            }

            history.append(record)

            print(
                f"{model_name} | "
                f"Step {iteration:5d} | "
                f"Train: {record['train_loss']:.4f} | "
                f"Val: {val_loss:.4f} | "
                f"Time: {elapsed:.1f}s",
                flush=True
            )

    # -------------------------------------------------
    # FINAL EVALUATION
    # -------------------------------------------------

    final_val_loss = evaluate_model_scaling(model)

    torch.cuda.synchronize()

    total_time = time.perf_counter() - start_time

    peak_memory = (
        torch.cuda.max_memory_allocated() / 1e9
    )

    final_train_loss = float(
        np.mean(recent_losses[-100:])
    )

    # Record final result
    final_record = {
        "iteration": iteration,
        "train_loss": final_train_loss,
        "val_loss": final_val_loss,
        "elapsed_seconds": total_time
    }

    history.append(final_record)

    scaling_histories[model_name] = history

    # -------------------------------------------------
    # SAVE CHECKPOINT
    # -------------------------------------------------

    torch.save({
        "model_state_dict": model.state_dict(),
        "config": config,
        "parameters": n_params,
        "iterations": iteration,
        "training_tokens": len(training_data),
        "validation_loss": final_val_loss,
        "seed": SEED,
        "history": history
    }, f"checkpoints/model_scaling/{model_name}.pt")

    scaling_summary.append({
        "Model": model_name,
        "Parameters": n_params,
        "Iterations": iteration,
        "Tokens processed": iteration * BATCH_SIZE * BLOCK_SIZE,
        "Final train loss": final_train_loss,
        "Final val loss": final_val_loss,
        "Val perplexity": math.exp(final_val_loss),
        "Training time (s)": total_time,
        "Peak GPU memory (GB)": peak_memory
    })

    print("\nCompleted:", model_name)
    print("Iterations:", iteration)
    print("Validation loss:", round(final_val_loss, 5))
    print("Training time:", round(total_time, 2), "seconds")
    print("Peak memory:", round(peak_memory, 2), "GB")

    del model, optimizer, scaler
    gc.collect()
    torch.cuda.empty_cache()

    # Save results after each model
    pd.DataFrame(scaling_summary).to_csv(
        "artifacts/model_scaling/results.csv",
        index=False
    )

# -----------------------------------------------------
# 5. FINAL COMPARISON
# -----------------------------------------------------

scaling_df = pd.DataFrame(scaling_summary)

print("\n" + "=" * 70)
print("EXPERIMENT E3: MODEL SCALING RESULTS")
print("=" * 70)

display(scaling_df)

scaling_df.to_csv(
    "artifacts/model_scaling/results.csv",
    index=False
)

print("\nE3 completed. Checkpoints and results saved.")
```
